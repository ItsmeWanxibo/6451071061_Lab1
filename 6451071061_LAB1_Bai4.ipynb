# LAB1 - Bài 4 (Titanic): Cổng lên tàu, loại vé và khả năng sống sót (theo giới tính)

**Họ tên:** Đinh Công Vinh Quang  
**MSSV:** 6451071061

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")

HO_TEN, MSSV = "Đinh Công Vinh Quang", "6451071061"
print(f"{HO_TEN} - {MSSV}")
```

```python
df = pd.read_csv("titanic.csv")
data = df.dropna(subset=["embarked"]).copy()
print(data.groupby(["embarked", "sex"])["survived"].mean().unstack().round(3), "\n")
print(data.groupby(["pclass", "sex"])["survived"].mean().unstack().round(3))
```

```python
g = sns.catplot(data=data, x="embarked", y="survived", hue="sex", kind="bar",
                order=["C", "Q", "S"], height=4, aspect=1.3, palette="Set2")
g.set_axis_labels("Cổng lên tàu", "Tỉ lệ sống sót")
g.figure.suptitle(f"Cổng lên tàu | {HO_TEN} - {MSSV}", y=1.03)
plt.show()
```

```python
g = sns.catplot(data=data, x="pclass", y="survived", hue="sex", kind="bar",
                height=4, aspect=1.3, palette="Set2")
g.set_axis_labels("Loại vé (pclass)", "Tỉ lệ sống sót")
g.figure.suptitle(f"Loại vé | {HO_TEN} - {MSSV}", y=1.03)
plt.show()
```

**Nhận xét:**
- Hành khách lên tàu ở cổng **C** có tỉ lệ sống sót cao nhất (~56%) so với Q (~36%) và S (~33%). Sự khác biệt này đúng ở cả hai giới: nữ cổng C ~90% (Q ~62%, S ~68%), nam cổng C ~31% (Q ~11%, S ~17%).
- Nguyên nhân có thể nằm ở cơ cấu hành khách: cổng C có ~52% khách hạng 1 (cổng Q chỉ ~2%, cổng S ~19%), nên cổng lên tàu liên quan gián tiếp qua loại vé.
- Loại vé liên quan rất mạnh: hạng 1 sống sót nhiều nhất (~62%), hạng 2 (~43%), hạng 3 thấp nhất (~26%). Ở cả nam và nữ, hạng vé càng cao thì tỉ lệ sống sót càng cao; nữ hạng 3 chỉ còn ~50%.
