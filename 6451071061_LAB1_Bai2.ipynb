# LAB1 - Bài 2 (Titanic): 5 thuộc tính có nhiều giá trị rỗng nhất

**Họ tên:** Đinh Công Vinh Quang  
**MSSV:** 6451071061

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")

HO_TEN, MSSV = "Đinh Công Vinh Quang", "6451071061"
print(f"{HO_TEN} - {MSSV}")
```

```python
df = pd.read_csv("titanic.csv")
null_count = df.isnull().sum().sort_values(ascending=False)
null_pct = (null_count / len(df) * 100).round(1)
top5 = pd.DataFrame({"so_gia_tri_rong": null_count, "ty_le_%": null_pct}).head(5)
top5
```

```python
top5["ty_le_%"].plot(kind="barh", color="tomato", figsize=(7, 3))
plt.xlabel("% giá trị rỗng"); plt.title(f"Top 5 thuộc tính thiếu dữ liệu - {HO_TEN} - {MSSV}")
plt.gca().invert_yaxis(); plt.show()
```

**Nhận xét:** thuộc tính `body` thiếu nghiêm trọng nhất (~90.8%), tiếp theo là `cabin` (~77.5%), `boat` (~62.9%), `home.dest` (~43.1%) và `age` (~20.1%). Ba thuộc tính đầu thiếu quá nhiều nên không nên dùng trực tiếp để khai thác; riêng `age` thiếu ít hơn nên có thể điền giá trị thiếu.
